```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from itertools import combinations

def binary_image_from_path(image_path, threshold=128):
    """Convert an image to a binary image."""
    image = Image.open(image_path).convert("L")  # Convert to grayscale
    binary_image = np.array(image) > threshold  # Binary threshold
    return binary_image.astype(int)

def generate_subsets(k):
    """Generate all subsets of even and odd cardinality."""
    elements = list(range(k))
    even_subsets = [set(comb) for r in range(0, k + 1, 2) for comb in combinations(elements, r)]
    odd_subsets = [set(comb) for r in range(1, k + 1, 2) for comb in combinations(elements, r)]
    return even_subsets, odd_subsets

def construct_matrices(k):
    """Construct C0 and C1 matrices using subsets."""
    even_subsets, odd_subsets = generate_subsets(k)
    m = 2**(k-1)
    alpha = 1 / m
    d = int(m * (1 - alpha))

    C0 = np.zeros((k, m), dtype=int)
    C1 = np.zeros((k, m), dtype=int)

    # Construct the matrices
    for i in range(k):
        for j, subset in enumerate(even_subsets):
            if i in subset:
                C0[i, j] = 1
        for j, subset in enumerate(odd_subsets):
            if i in subset:
                C1[i, j] = 1

    return C0, C1, m, alpha, d

def apply_conditions(subpixel_pattern, k, m, alpha, d, pixel_value):
    """Apply contrast, security, and uniformity conditions."""
    # Step 1: Apply security with random permutations
    permuted_pattern = subpixel_pattern[:, np.random.permutation(m)]

    # Step 2: Calculate Hamming weight for contrast
    combined_hamming_weight = np.sum(np.any(permuted_pattern, axis=0))

    # Step 3: Check contrast conditions
    if pixel_value == 0:  # Black pixel
        if combined_hamming_weight < d:
            return permuted_pattern
    else:  # White pixel
        if combined_hamming_weight >= d - alpha * m:
            return permuted_pattern

    # Step 4: Retry until conditions are satisfied
    for _ in range(100):  # Limit retries
        permuted_pattern = subpixel_pattern[:, np.random.permutation(m)]
        combined_hamming_weight = np.sum(np.any(permuted_pattern, axis=0))

        if (pixel_value == 0 and combined_hamming_weight >= d) or \
           (pixel_value == 1 and combined_hamming_weight < d - alpha * m):
            return permuted_pattern

    raise ValueError("Failed to satisfy the contrast condition.")

def construct_shares(image, k):
    """Generate shares using Construction 2 with applied conditions."""
    height, width = image.shape
    C0, C1, m, alpha, d = construct_matrices(k)
    shares = np.zeros((k, height, width * m), dtype=int)

    # Generate shares for each pixel
    for i in range(height):
        for j in range(width):
            pixel = image[i, j]
            subpixel_pattern = C0 if pixel == 0 else C1

            # Apply all conditions
            valid_pattern = apply_conditions(subpixel_pattern, k, m, alpha, d, pixel)

            for s in range(k):
                shares[s, i, j * m : (j + 1) * m] = valid_pattern[s]

    return shares, m, alpha, d

def reconstruct_image(shares, k, m, alpha, d):
    """Reconstruct the image using the logical OR of shares."""
    height, full_width = shares[0].shape
    width = full_width // m

    reconstructed = np.zeros((height, width), dtype=int)
    for i in range(height):
        for j in range(width):
            subpixel_sum = np.zeros(m, dtype=int)
            for s in range(k):
                subpixel_sum |= shares[s, i, j * m : (j + 1) * m]
            reconstructed[i, j] = 1 if np.sum(subpixel_sum) >= d else 0
    return reconstructed

def display_image(image, title):
    """Display a single image with a title."""
    plt.figure(figsize=(5, 5))
    plt.imshow(image, cmap="gray")
    plt.title(title)
    plt.axis("off")
    plt.show()

# Parameters
k = 3
image_path = "/home/arvind/Visual-Cryptography/Alphabets/A.jpeg"

# Step 1: Convert image to binary
binary_image = binary_image_from_path(image_path)

# Step 2: Generate shares
shares, m, alpha, d = construct_shares(binary_image, k)

# Step 3: Reconstruct the image
reconstructed_image = reconstruct_image(shares, k, m, alpha, d)

# Step 4: Display results
print(f"Alpha: {alpha}, m: {m}, d: {d}")
display_image(binary_image, "Original Binary Image")
for i in range(k):
    display_image(shares[i], f"Share {i+1}")
display_image(reconstructed_image, "Reconstructed Image")
```
